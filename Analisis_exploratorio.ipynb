ANALISIS EXPLORATORIO BASE DE DATOS MINAS

```python
#Importar la base de datos y ver las primeras 5 Lineas 
import pandas as pd
data = pd.read_csv('/Users/franck/Documents/GitHub/ProyectoData/Eventos_Minas_Antipersonal_en_Colombia_20251027.csv')
display(data.head())
```

```python
#Hacer un analisis de las columnas y su tipo de dato
display(data.info())
```

```python
#Eliminar los valores nulos
data_cleaned = data.dropna()
display(data_cleaned.info())
```

```python
# Graficar las distribuciones de las variables numéricas excepto latitud y longitud
import matplotlib.pyplot as plt
data_cleaned.hist(bins=30, figsize=(15, 10))
plt.tight_layout()
plt.show()
```

```python
#Graficar solo la variable TIPO_EVENTO
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10, 5))
sns.countplot(data=data_cleaned, x='TIPO_EVENTO', order=data_cleaned['TIPO_EVENTO'].value_counts().index)
plt.title('Distribución de TIPO_EVENTO')
plt.xticks(rotation=45)
plt.show()
```

```python
#Graficar solo la variable DEPARTAMENTO
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10, 5))
sns.countplot(data=data_cleaned, x='DEPARTAMENTO', order=data_cleaned['DEPARTAMENTO'].value_counts().index)
plt.title('Distribución de DEPARTAMENTO')
plt.xticks(rotation=45)
plt.show()
```

```python
#Graficar solo la variable EVENTO
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10, 5))
sns.countplot(data=data_cleaned, x='EVENTO', order=data_cleaned['EVENTO'].value_counts().index)
plt.title('Distribución de EVENTO')
plt.xticks(rotation=45)
plt.show()
```
